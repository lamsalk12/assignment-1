```python
import pandas as pd
import numpy as np

DATA_PATH = "../data/scores_raw.csv"
OUTPUT_PATH = "../outputs/summary.csv"
```

```python
# C1 — Series and DataFrame by hand
marks_series = pd.Series(
    [88, 74, 91, 60, 79],
    index=["Aarav", "Bibek", "Chandni", "Deepika", "Esha"],
)
print("C1 — Series index:", marks_series.index.tolist())
print("C1 — Series values:", marks_series.values)

# Same small table, built two different ways.
data_dict = {
    "name": ["Aarav", "Bibek", "Chandni", "Deepika", "Esha"],
    "marks": [88, 74, 91, 60, 79],
}
df_from_dict = pd.DataFrame(data_dict)

records = [
    {"name": "Aarav", "marks": 88},
    {"name": "Bibek", "marks": 74},
    {"name": "Chandni", "marks": 91},
    {"name": "Deepika", "marks": 60},
    {"name": "Esha", "marks": 79},
]
df_from_records = pd.DataFrame(records)

print("C1 — dict-of-lists equals list-of-dicts:", df_from_dict.equals(df_from_records))
print("C1 — shape:", df_from_dict.shape)
print("C1 — columns:", df_from_dict.columns.tolist())
print("C1 — index:", df_from_dict.index.tolist())
```

```python
#C2 — Load and inspect
df_raw = pd.read_csv(DATA_PATH)
df = df_raw.copy()

print("C2 — head():")
print(df.head())
```

```python
print("C2 — tail(3):")
print(df.tail(3))
```

```python
print("C2 — shape:", df.shape)
```

```python
print("C2 — info():")
df.info()
```

```python
print("C2 — describe():")
print(df.describe())
```

```python
print("C2 — describe(include='object'):")
print(df.describe(include="str"))
```

```python
#C3 — Selecting rows and columns
col_series = df["marks"]
col_frame = df[["marks"]]
print("C3 — type(df['marks']):", type(col_series))
print("C3 — type(df[['marks']]):", type(col_frame))
```

```python
loc_result = df.loc[0:4, ["name", "subject", "marks"]]
iloc_result = df.iloc[0:5, 1:4]
print("C3 — loc rows returned:", len(loc_result))
print("C3 — iloc rows returned:", len(iloc_result))
```

```python
#C4 — Boolean filtering
df_c4a = df[(df["marks"] > 80) & (df["subject"] == "Python Programming")]
print("C4a — marks > 80 and subject == Python Programming:", len(df_c4a), "rows")
```

```python
df_c4b_dirty = df[df["city"].isin(["Kathmandu", "Pokhara"])]
print("C4b — rows from Kathmandu or Pokhara(without clean):", len(df_c4b_dirty))
```

```python
df_c4c = df.query("marks > 80 and subject == 'Python Programming'")
print("C4c — same filter via .query(), identical to C4a:", df_c4a.equals(df_c4c))
```

```python
#C5 — Cleaning inconsistent text
print("C5 — unique city spellings before cleaning:", df["city"].nunique())
print(df["city"].value_counts())
```

```python
df["city"] = df["city"].str.strip().str.title()

print("C5 — unique city spellings after cleaning:", df["city"].nunique())
print(df["city"].value_counts())
```

```python
df_c5b = df[df["city"].isin(["Kathmandu", "Pokhara"])]
rows_recovered = len(df_c5b) - len(df_c4b_dirty)
print(f"C5 — C4b filter now returns {len(df_c5b)} rows (was {len(df_c4b_dirty)}); "
      f"{rows_recovered} rows were being lost to inconsistent spelling.")
```

```python
#C6 — Missing values
print("C6 — missing values per column:")
print(df.isna().sum())
```

```python
df_c6_demo = df.copy()

before_fill = df_c6_demo["attendance_percent"].isna().sum()
df_c6_demo["attendance_percent"] = df_c6_demo["attendance_percent"].fillna(
    df_c6_demo["attendance_percent"].mean()
)
print(f"C6 — attendance_percent missing before fill: {before_fill}, "
      f"after fill: {df_c6_demo['attendance_percent'].isna().sum()}")
```

```python
before_drop = len(df_c6_demo)
df_c6_demo = df_c6_demo.dropna(subset=["marks"])
print(f"C6 — rows before dropping missing marks: {before_drop}, after: {len(df_c6_demo)}")
```

```python
#C7 — Duplicates
print("C7 — fully duplicated rows:", df.duplicated().sum())
print(df[df.duplicated()].head())
```

```python
before_dedup = df.shape[0]
df = df.drop_duplicates(keep="first")
print(f"C7 — shape before: {before_dedup} rows, after: {df.shape[0]} rows")
```

```python
dup_subset = df.duplicated(subset=["student_id", "subject"]).sum()
print("C7 — duplicates on ['student_id', 'subject'] subset:", dup_subset)
```

```python
df["attendance_percent"] = df["attendance_percent"].fillna(df["attendance_percent"].mean())
before_drop2 = len(df)
df = df.dropna(subset=["marks"])
print(f"Applying C6's cleaning after dedup: {before_drop2} -> {len(df)} rows")
```

```python
#C8 — Dates
print("C8 — exam_date dtype before:", df["exam_date"].dtype)
```

```python
df["exam_date"] = pd.to_datetime(df["exam_date"])
print("C8 — exam_date dtype after:", df["exam_date"].dtype)
```

```python
df["exam_month"] = df["exam_date"].dt.month
df["exam_weekday"] = df["exam_date"].dt.day_name()

after_cutoff = df[df["exam_date"] > "2026-02-15"]
print("C8 — exams held after 2026-02-15:", len(after_cutoff))
```

```python
#C9 — Sorting and renaming 
sorted_df = df.sort_values(["subject", "marks"], ascending=[True, False])
print("C9 — sorted by subject (A-Z), then marks (high to low):")
print(sorted_df.head(10)[["name", "subject", "marks"]])
```

```python
renamed_df = df.rename(columns={"attendance_percent": "attendance"})
print("C9 — renamed_df columns:", renamed_df.columns.tolist())
print("C9 — original df columns unchanged:", df.columns.tolist())
```

```python
restored_df = sorted_df.sort_index()
print("C9 — restored to original row order, first 5 index labels:", restored_df.index[:5].tolist())
```

```python
#C10 — Derived columns
n_impossible = (df["marks"] > 100).sum()
print("C10 — impossible marks (>100) found:", n_impossible)
```

```python
df = df[df["marks"] <= 100]
print("C10 — rows after dropping impossible marks:", len(df))

df["passed"] = df["marks"] >= 40
df["grade"] = np.where(
    df["marks"] >= 85, "A",
    np.where(df["marks"] >= 70, "B",
             np.where(df["marks"] >= 40, "C", "F"))
)
```

```python
pass_rate = df["passed"].astype(int).mean() * 100
print(f"C10 — overall pass rate: {pass_rate:.1f}%")
print("C10 — grade value counts:")
print(df["grade"].value_counts())
```

```python
#C11 — Types and memory
before_mem = df[["subject", "city"]].memory_usage(deep=True, index=False)
print("C11 — memory before (bytes):")
print(before_mem)
```

```python
df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")

after_mem = df[["subject", "city"]].memory_usage(deep=True, index=False)
print("C11 — memory after (bytes):")
print(after_mem)
```

```python
pct_saved = (1 - after_mem.sum() / before_mem.sum()) * 100
print(f"C11 — percentage saved: {pct_saved:.1f}%")
```

```python
#C12 — Aggregation and a summary table
agg_by_subject = df.groupby("subject", observed=True)["marks"].agg(["mean", "max", "count"])
print("C12a — mean/max/count of marks per subject:")
print(agg_by_subject)
```

```python
mean_by_city_subject = df.groupby(["city", "subject"], observed=True)["marks"].mean()
print("C12b — mean marks by city and subject:")
print(mean_by_city_subject)
```

```python
pivot = df.pivot_table(values="marks", index="city", columns="subject", aggfunc="mean")
print("C12c — pivot table, mean marks (rows=city, columns=subject):")
print(pivot)
```

```python
mean_by_month = df.groupby("exam_month")["marks"].mean().sort_index()
print("C12d — mean marks per exam month (chronological):")
print(mean_by_month)
```

```python
agg_by_subject.to_csv(OUTPUT_PATH)
print(f"C12 — exported (a) to {OUTPUT_PATH}")
```
