```python
import numpy as np
import time
```

```python
np.random.seed(7)
cities = np.array(["Kathmandu", "Pokhara", "Biratnagar", "Lalitpur", "Birgunj"])
days = np.array([f"D{i:02d}" for i in range(1, 15)])
```

```python
temps = np.random.randint(14, 36, size=(5, 14)).astype(float)
humidity = np.random.randint(40, 95, size=(5, 14)).astype(float)
```

```python
temps[2, 5] = np.nan
temps[4, 11] = np.nan
```

```python
#B1 — Array anatomy and reshaping
print("B1a — temps.shape:", temps.shape)
print("B1a — temps.ndim:", temps.ndim)
print("B1a — temps.dtype:", temps.dtype)
print("B1a — temps.size:", temps.size)
```

```python
flat70 = np.arange(70)
reshaped_explicit = flat70.reshape(5, 14)
reshaped_neg1 = flat70.reshape(5, -1)
print("B1b — reshape(5,14) == reshape(5,-1):", np.array_equal(reshaped_explicit, reshaped_neg1))
back_to_flat = reshaped_explicit.flatten()
print("B1b — flattened back matches original:", np.array_equal(back_to_flat, flat70))
```

```python
#B5 — Vectorization vs. a Python loop
def to_fahrenheit_loop(arr):
    result = np.empty_like(arr)
    flat_in = arr.flatten()
    flat_out = result.flatten()
    for i in range(flat_in.size):
        flat_out[i] = flat_in[i] * 9 / 5 + 32
    return flat_out.reshape(arr.shape)
 
fahrenheit_loop = to_fahrenheit_loop(temps)
fahrenheit_vec = temps * 9 / 5 + 32
print("B5 — loop and vectorized agree:", np.allclose(fahrenheit_loop, fahrenheit_vec, equal_nan=True))
```

```python
big = np.random.uniform(14, 36, size=1_000_000)
 
start = time.perf_counter()
big_loop_result = np.empty_like(big)
for i in range(big.size):
    big_loop_result[i] = big[i] * 9 / 5 + 32
loop_time = time.perf_counter() - start
 
start = time.perf_counter()
big_vec_result = big * 9 / 5 + 32
vec_time = time.perf_counter() - start
 
speedup = loop_time / vec_time
print(f"B5 — loop time: {loop_time:.4f}s")
print(f"B5 — vectorized time: {vec_time:.4f}s")
print(f"B5 — speedup: ~{speedup:.1f}x faster")
```

```python
#B7 — Matrix multiplication
quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])
print("B7a — element-wise (*):", quantities * prices)
print("B7a — dot product (@):", quantities @ prices)
```

```python
#B8 — Aggregating along an axis
features = np.stack([
    np.nanmean(temps, axis=0),
    np.nanmean(humidity, axis=0)
], axis=1)
weights = np.array([0.6, -0.1])
comfort_score = features @ weights
print("B7b — features shape:", features.shape)
print("B7b — weights shape:", weights.shape)
print("B7b — comfort_score shape:", comfort_score.shape)
```

```python
#B9 — Reproducible randomness and a train/test split
flat_samples = temps.flatten()
n = flat_samples.size
 
np.random.seed(42)
idx = np.arange(n)
np.random.shuffle(idx)
split_point = int(n * 0.8)
train_idx, test_idx = idx[:split_point], idx[split_point:]
print("B9 — train shape:", train_idx.shape, "| test shape:", test_idx.shape)
print("B9 — sizes sum to 70:", train_idx.size + test_idx.size == 70)
print("B9 — no shared indices:", np.intersect1d(train_idx, test_idx).size == 0)
```

```python
np.random.seed(42)
idx2 = np.arange(n)
np.random.shuffle(idx2)
train_idx2, test_idx2 = idx2[:split_point], idx2[split_point:]
assert np.array_equal(train_idx, train_idx2) and np.array_equal(test_idx, test_idx2)
print("B9 — repeated split with same seed is identical: True")
```
